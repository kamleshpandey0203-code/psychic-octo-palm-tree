### Installing Python Libraries

In Google Colab, you can install Python libraries using the `pip` package manager. You need to prefix the `pip` command with an exclamation mark `!` to run it as a shell command within the notebook environment.

```python
# To install a library, use the following syntax:
# !pip install <library_name>

# For example, to install the pandas library:
!pip install pandas
```

After installation, you can import and use the library in subsequent cells.
